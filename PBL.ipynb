```python
import os
from typing import List, Literal, Optional

LLMProvider = Literal["openai", "gemini"]

class LLMClient:
    def __init__(self) -> None:
        self.provider: Optional[LLMProvider] = None
        self._openai_client = None
        self._gemini_model = None
        provider = os.getenv("LLM_PROVIDER", "").lower()
        if provider in {"openai", "gemini"}:
            self.provider = provider  # type: ignore[assignment]
        if self.provider == "openai":
            self._init_openai()
        elif self.provider == "gemini":
            self._init_gemini()

    def _init_openai(self) -> None:
        try:
            import openai  # type: ignore[import]
        except Exception:
            self.provider = None
            return
        api_key = os.getenv("OPENAI_API_KEY")
        if not api_key:
            self.provider = None
            return
        openai.api_key = api_key
        self._openai_client = openai

    def _init_gemini(self) -> None:
        try:
            import google.generativeai as genai  # type: ignore[import]
        except Exception:
            self.provider = None
            return
        api_key = os.getenv("GEMINI_API_KEY")
        if not api_key:
            self.provider = None
            return
        genai.configure(api_key=api_key)
        model_name = os.getenv("GEMINI_MODEL", "gemini-1.5-flash")
        self._gemini_model = genai.GenerativeModel(model_name)

    def legal_chat_answer(self, question: str, context_snippets: List[str]) -> str:
        question = (question or "").strip()
        joined = "\n\n".join(context_snippets[:5])
        prompt = (
            "You are an AI legal information assistant for ordinary citizens. "
            "You are not a lawyer and cannot give official legal advice. "
            "Using ONLY the information in the context and your public knowledge, "
            "answer the user's question in clear, non-technical language, 3–6 short paragraphs.\n\n"
            f"User question:\n{question}\n\n"
            f"Context snippets:\n{joined}\n\n"
            "If you are unsure, say that they should consult a qualified lawyer or legal aid clinic."
        )
        out = self._generate_text(prompt)
        if out:
            return out.strip()
        return (
            "This is general legal information based on similar situations. "
            "For decisions about your specific case, please speak with a qualified lawyer or free legal aid clinic."
        )

    def simplify_clause(self, clause: str) -> str:
        clause = (clause or "").strip()
        if not clause:
            return "I need the clause text to explain it simply."
        prompt = (
            "Explain the following legal clause in very simple language for a 10 year old. "
            "Avoid legal jargon. Keep it under 3 sentences.\n\n"
            f"Clause:\n{clause}"
        )
        out = self._generate_text(prompt)
        if out:
            return out.strip()
        return (
            "This clause talks about who has which responsibilities. "
            "If something goes wrong, it explains who might have to fix the problem or pay for it."
        )

    def multilingual_explanation(self, text: str, language: str) -> str:
        text = (text or "").strip()
        if not text:
            return "No explanation to translate."
        prompt = (
            "Translate the following explanation for ordinary citizens into the "
            f"{language} language. Keep the tone clear, respectful, and simple.\n\n"
            f"Text:\n{text}"
        )
        out = self._generate_text(prompt)
        if out:
            return out.strip()
        return f"[Translation to {language} not available in offline mode]\n\n{text}"

    def _generate_text(self, prompt: str) -> str:
        if not self.provider:
            return ""
        if self.provider == "openai" and self._openai_client:
            try:
                res = self._openai_client.ChatCompletion.create(
                    model=os.getenv("OPENAI_MODEL", "gpt-4o-mini"),
                    messages=[
                        {"role": "system", "content": "You are a helpful legal information assistant."},
                        {"role": "user", "content": prompt},
                    ],
                    temperature=0.3,
                )
                return res.choices[0].message["content"]
            except Exception:
                return ""
        if self.provider == "gemini" and self._gemini_model:
            try:
                res = self._gemini_model.generate_content(prompt)
                return getattr(res, "text", "") or ""
            except Exception:
                return ""
        return ""

llm_client = LLMClient()
```

```python
write(f"{ROOT}/backend/vectify_client.py", r"""
from __future__ import annotations
import os
from typing import List
import requests
VECTIFY_API_URL = os.getenv("VECTIFY_API_URL", "").rstrip("/")
VECTIFY_API_KEY = os.getenv("VECTIFY_API_KEY", "")
def retrieve_legal_context(query: str, top_k: int = 5) -> List[str]:
    '''
    Call VectifyAI to retrieve legal context (e.g. from GitHub repositories).
    This is a placeholder that expects a generic VectifyAI REST API:
    - POST {VECTIFY_API_URL}/retrieve
      { "query": "...", "top_k": 5 }
    Adjust the URL / body to match your VectifyAI project.
    '''
    if not VECTIFY_API_URL or not VECTIFY_API_KEY:
        # Offline/demo fallback
        return [
            "VectifyAI is not configured. Please set VECTIFY_API_URL and VECTIFY_API_KEY in the environment.",
            "In production, this will return legal context snippets from your GitHub-connected VectifyAI project.",
        ]
    try:
        resp = requests.post(
            f"{VECTIFY_API_URL}/retrieve",
            json={"query": query, "top_k": top_k},
            headers={"Authorization": f"Bearer {VECTIFY_API_KEY}"},
            timeout=20,
        )
        resp.raise_for_status()
        data = resp.json()
        snippets = []
        for item in data.get("results", []):
            text = item.get("text") or item.get("content") or ""
            if text:
                snippets.append(text)
        return snippets or ["No relevant legal context found by VectifyAI."]
    except Exception as e:
        return [f"VectifyAI retrieval failed: {e}"]
""")
```

```python
write(f"{ROOT}/backend/legal_policy_model.py", r'''
from __future__ import annotations
from typing import Dict
import torch
from transformers import AutoModelForSequenceClassification, AutoTokenizer
_MODEL_NAME = "nlpaueb/legal-bert-base-uncased"
_NUM_LABELS = 4  # SAFE, MODERATE_RISK, HIGH_RISK, MISSING_CLAUSE
_label_map = {
    0: "SAFE",
    1: "MODERATE_RISK",
    2: "HIGH_RISK",
    3: "MISSING_CLAUSE",
}
_tokenizer = None
_model = None
def _load_model():
    global _tokenizer, _model
    if _tokenizer is not None and _model is not None:
        return _tokenizer, _model
    try:
        _tokenizer = AutoTokenizer.from_pretrained(_MODEL_NAME)
        _model = AutoModelForSequenceClassification.from_pretrained(_MODEL_NAME, num_labels=_NUM_LABELS)
        _model.eval()
    except Exception:
        _tokenizer = None
        _model = None
    return _tokenizer, _model
def score_clause(clause: str) -> Dict[str, float | str]:
    """
    Given a clause, return:
    - risk_score (float 0-1)
    - risk_class (SAFE, MODERATE_RISK, HIGH_RISK, MISSING_CLAUSE)
    """
    clause = (clause or "").strip()
    if not clause:
        return {"risk_score": 0.0, "risk_class": "MISSING_CLAUSE"}
    tokenizer, model = _load_model()
    if tokenizer is None or model is None:
        # Fallback heuristic: longer clauses slightly higher risk
        length = len(clause.split())
        if length < 10:
            return {"risk_score": 0.1, "risk_class": "SAFE"}
        elif length < 40:
            return {"risk_score": 0.4, "risk_class": "MODERATE_RISK"}
        else:
            return {"risk_score": 0.7, "risk_class": "HIGH_RISK"}
    with torch.no_grad():
        inputs = tokenizer(
            clause,
            truncation=True,
            padding="max_length",
            max_length=256,
            return_tensors="pt",
        )
        outputs = model(**inputs)
        probs = outputs.logits.softmax(dim=-1)[0]
        score, idx = float(probs.max().item()), int(probs.argmax().item())
        risk_class = _label_map.get(idx, "SAFE")
        return {"risk_score": score, "risk_class": risk_class}
''')
```

scam_policy_model.py (DistilBERT RL policy stub)

```python
write(f"{ROOT}/backend/scam_policy_model.py", r'''
from __future__ import annotations
from typing import Dict
import torch
from transformers import AutoModelForSequenceClassification, AutoTokenizer
_MODEL_NAME = "distilbert-base-uncased"
_NUM_LABELS = 4  # NOT_SCAM, SCAM_GENERIC, CREDENTIAL_HARVEST, MONEY_DEMAND
_label_to_type = {
    0: "NOT_SCAM",
    1: "GENERIC_SCAM",
    2: "CREDENTIAL_HARVESTING",
    3: "MONEY_DEMAND",
}
_tokenizer = None
_model = None
def _load_model():
    global _tokenizer, _model
    if _tokenizer is not None and _model is not None:
        return _tokenizer, _model
    try:
        _tokenizer = AutoTokenizer.from_pretrained(_MODEL_NAME)
        _model = AutoModelForSequenceClassification.from_pretrained(_MODEL_NAME, num_labels=_NUM_LABELS)
        _model.eval()
    except Exception:
        _tokenizer = None
        _model = None
    return _tokenizer, _model
def score_message(text: str) -> Dict[str, str | float]:
    """
    Return:
    - scam_probability (0-1)
    - phishing_type
    - legal_section (rough mapping)
    """
    text = (text or "").strip()
    if not text:
        return {
            "scam_probability": 0.0,
            "phishing_type": "NOT_SCAM",
            "legal_section": "NONE",
        }
    tokenizer, model = _load_model()
    if tokenizer is None or model is None:
        # Simple heuristic fallback
        lower = text.lower()
        prob = 0.0
        ptype = "NOT_SCAM"
        if "otp" in lower or "password" in lower or "click the link" in lower:
            prob = 0.9
            ptype = "CREDENTIAL_HARVESTING"
        elif "lottery" in lower or "prize" in lower or "won" in lower:
            prob = 0.8
            ptype = "GENERIC_SCAM"
        elif "transfer" in lower and "money" in lower:
            prob = 0.75
            ptype = "MONEY_DEMAND"
        legal = "IT_ACT_66D" if prob >= 0.5 else "NONE"
        return {
            "scam_probability": prob,
            "phishing_type": ptype,
            "legal_section": legal,
        }
    with torch.no_grad():
        inputs = tokenizer(
            text,
            truncation=True,
            padding="max_length",
            max_length=256,
            return_tensors="pt",
        )
        outputs = model(**inputs)
        probs = outputs.logits.softmax(dim=-1)[0]
        prob, idx = float(probs[1:].max().item()), int(probs[1:].argmax().item()) + 1
        phishing_type = _label_to_type.get(idx, "GENERIC_SCAM")
        legal_section = "IT_ACT_66D" if prob >= 0.5 else "NONE"
        return {
            "scam_probability": prob,
            "phishing_type": phishing_type,
            "legal_section": legal,
        }
''')
```

document_analysis.py (OCR + Legal-BERT RL scoring + LLM explain)

```python
write(f"{ROOT}/backend/document_analysis.py", r"""
from __future__ import annotations
import io
from dataclasses import dataclass
from typing import List, Optional, Tuple
from fastapi import UploadFile
from legal_policy_model import score_clause
from llm_client import llm_client
from utils import safe_str
try:
    import pdfplumber  # type: ignore[import]
except Exception:
    pdfplumber = None  # type: ignore[assignment]
try:
    from PIL import Image  # type: ignore[import]
    import pytesseract  # type: ignore[import]
except Exception:
    Image = None  # type: ignore[assignment]
    pytesseract = None  # type: ignore[assignment]
try:
    import spacy  # type: ignore[import]
except Exception:
    spacy = None  # type: ignore[assignment]
_NLP = None
def _get_nlp():
    global _NLP
    if _NLP is not None:
        return _NLP
    if spacy is None:
        return None
    try:
        _NLP = spacy.load("en_core_web_sm")
    except Exception:
        _NLP = None
    return _NLP
@dataclass
class ClauseRisk:
    clause: str
    risk_score: float
    risk_class: str
    explanation: Optional[str] = None
@dataclass
class ContractAnalysisResult:
    text: str
    clauses: List[ClauseRisk]
    safety_score: int
    safety_level: str
def _extract_text_from_pdf(data: bytes) -> str:
    if pdfplumber is None:
        return ""
    text_parts: List[str] = []
    with pdfplumber.open(io.BytesIO(data)) as pdf:
        for page in pdf.pages:
            t = page.extract_text() or ""
            if t.strip():
                text_parts.append(t)
    return "\n".join(text_parts)
def _extract_text_from_image(data: bytes) -> str:
    if Image is None or pytesseract is None:
        return ""
    img = Image.open(io.BytesIO(data))
    return pytesseract.image_to_string(img)
async def extract_text_from_upload(file: UploadFile) -> str:
    data = await file.read()
    suffix = (file.filename or "").lower()
    if suffix.endswith(".pdf"):
        return safe_str(_extract_text_from_pdf(data))
    if any(suffix.endswith(ext) for ext in [".png", ".jpg", ".jpeg", ".tif", ".tiff"]):
        return safe_str(_extract_text_from_image(data))
    try:
        return safe_str(data.decode("utf-8", errors="ignore"))
    except Exception:
        return ""
def segment_clauses(text: str) -> List[str]:
    text = safe_str(text)
    if not text:
        return []
    nlp = _get_nlp()
    if not nlp:
        raw = [p.strip() for p in text.replace("\r", "\n").split("\n") if p.strip()]
        out: List[str] = []
        for para in raw:
            for s in para.split("."):
                s = s.strip()
                if len(s) > 20:
                    out.append(s)
        return out
    doc = nlp(text)
    return [sent.text.strip() for sent in doc.sents if len(sent.text.strip()) > 20]
def _safety_from_scores(risks: List[ClauseRisk]) -> Tuple[int, str]:
    # aggregate risk_score into simple 0–100 safety score
    if not risks:
        return 100, "Low"
    total_risk = 0.0
    for r in risks:
        if r.risk_class == "HIGH_RISK":
            total_risk += r.risk_score * 2.0
        elif r.risk_class == "MODERATE_RISK":
            total_risk += r.risk_score * 1.2
        elif r.risk_class == "MISSING_CLAUSE":
            total_risk += 0.8
        else:
            total_risk += r.risk_score * 0.5
    score = max(0.0, 100.0 - total_risk * 20.0)
    if score >= 80:
        level = "Low"
    elif score >= 60:
        level = "Moderate"
    elif score >= 40:
        level = "High"
    else:
        level = "Very High"
    return int(score), level
def analyze_contract_text(text: str) -> ContractAnalysisResult:
    clean = safe_str(text)
    clauses = segment_clauses(clean)
    clause_risks: List[ClauseRisk] = []
    for c in clauses:
        res = score_clause(c)
        risk_score = float(res.get("risk_score", 0.0))
        risk_class = str(res.get("risk_class", "SAFE"))
        explanation: Optional[str] = None
        if risk_class in {"MODERATE_RISK", "HIGH_RISK", "MISSING_CLAUSE"}:
            explanation = llm_client.simplify_clause(c)
        clause_risks.append(ClauseRisk(clause=c, risk_score=risk_score, risk_class=risk_class, explanation=explanation))
    safety_score, safety_level = _safety_from_scores(clause_risks)
    return ContractAnalysisResult(
        text=clean,
        clauses=clause_risks,
        safety_score=safety_score,
        safety_level=safety_level,
    )
""")
```

scam_analyzer.py (DistilBERT RL scoring)

```python
write(f"{ROOT}/backend/scam_analyzer.py", r"""
from __future__ import annotations
from dataclasses import dataclass
from typing import Dict
from llm_client import llm_client
from scam_policy_model import score_message
from utils import safe_str
@dataclass
class ScamAnalysis:
    scam_probability: float
    phishing_type: str
    legal_section: str
    explanation: str
def analyze_scam_message(message: str) -> ScamAnalysis:
    msg = safe_str(message)
    scores: Dict[str, str | float] = score_message(msg)
    prob = float(scores.get("scam_probability", 0.0))
    phishing_type = str(scores.get("phishing_type", "NOT_SCAM"))
    legal_section = str(scores.get("legal_section", "NONE"))
    # Optional LLM explanation
    explanation = ""
    base_text = (
        f"The following message may be a scam.\n\n"
        f"Message:\n{msg}\n\n"
        f"Model assessment: scam_probability={prob:.2f}, phishing_type={phishing_type}, legal_section={legal_section}.\n"
        "Explain this assessment in simple language and how the person should protect themselves."
    )
    if prob >= 0.5:
        explanation = llm_client._generate_text(base_text) or ""
    if not explanation:
        if prob < 0.3:
            explanation = "This message looks mostly safe, but you should still avoid sharing OTPs, passwords, or banking details."
        else:
            explanation = (
                "This message shows patterns similar to common online scams. "
                "Do not click unknown links or share OTPs, passwords, or bank information. "
                "Contact your bank or service provider directly using official channels."
            )
    return ScamAnalysis(
        scam_probability=prob,
        phishing_type=phishing_type,
        legal_section=legal_section,
        explanation=explanation,
    )
""")
```

```python
write(f"{ROOT}/backend/reporting.py", r"""
from __future__ import annotations
from typing import Dict, List
from document_analysis import ClauseRisk, ContractAnalysisResult
def build_report_payload(
    contract: ContractAnalysisResult,
    user_issue: str | None = None,
    suggested_actions: List[str] | None = None,
) -> Dict:
    risky_clauses: List[Dict] = []
    for c in contract.clauses:
        if c.risk_class in {"MODERATE_RISK", "HIGH_RISK", "MISSING_CLAUSE"}:
            risky_clauses.append(
                {
                    "clause": c.clause,
                    "risk_score": c.risk_score,
                    "risk_class": c.risk_class,
                    "explanation": c.explanation,
                }
            )
    return {
        "summary": {
            "safety_score": contract.safety_score,
            "safety_level": contract.safety_level,
            "risky_clause_count": len(risky_clauses),
        },
        "risky_clauses": risky_clauses,
        "user_issue": user_issue,
        "suggested_actions": suggested_actions or [],
    }
""")
```

```python
%%writefile {ROOT}/backend/complaint_generator.py
from __future__ import annotations
from dataclasses import dataclass
from typing import Literal
from llm_client import llm_client
from utils import safe_str
ComplaintType = Literal["consumer", "police", "legal_notice"]
@dataclass
class ComplaintRequest:
    complaint_type: ComplaintType
    user_name: str
    opponent_name: str
    facts: str
    location: str
def _base_template(req: ComplaintRequest) -> str:
    facts = safe_str(req.facts)
    return f'''
To,
The Concerned Authority
{req.location}
Subject: Complaint by {req.user_name} against {req.opponent_name}
Respected Sir/Madam,
I, {req.user_name}, wish to submit the following facts:
{facts}
I request that appropriate action be taken in accordance with law.
Yours faithfully,
{req.user_name}
'''.strip()
def generate_complaint(req: ComplaintRequest) -> str:
    base = _base_template(req)
    if req.complaint_type == "consumer":
        heading = "Consumer Complaint"
        extra = (
            "This complaint is being made in relation to deficiency of service / unfair trade practice. "
            "I request that the opposite party be directed to refund my money, pay compensation, and stop "
            "such practices."
        )
    elif req.complaint_type == "police":
        heading = "Police Complaint / FIR Information"
        extra = (
            "The above events may amount to a cognizable offence. I request that this information be "
            "treated as my complaint and that appropriate FIR / enquiry be initiated."
        )
    else:
        heading = "Legal Notice"
        extra = (
            "Through this notice, I call upon the opposite party to remedy the issue within 15 days, "
            "failing which I may take further legal action as advised."
        )
    draft = f"{heading}\n\n{base}\n\n{extra}"
    prompt = (
        "Rewrite the following draft as a clear, respectful legal complaint/notice suitable for an Indian "
        "context. Keep paragraphs short and easy to read. Do not invent new facts.\n\n"
        f"{draft}"
    )
    improved = llm_client._generate_text(prompt)  # type: ignore[attr-defined]
    return (improved or draft).strip()
```

```python
write(f"{ROOT}/backend/llm_client.py", r"""
from __future__ import annotations
import os
from typing import List, Literal, Optional

LLMProvider = Literal["openai", "gemini"]

class LLMClient:
    def __init__(self) -> None:
        self.provider: Optional[LLMProvider] = None
        self._openai_client = None
        self._gemini_model = None
        provider = os.getenv("LLM_PROVIDER", "").lower()
        if provider in {"openai", "gemini"}:
            self.provider = provider  # type: ignore[assignment]
        if self.provider == "openai":
            self._init_openai()
        elif self.provider == "gemini":
            self._init_gemini()

    def _init_openai(self) -> None:
        try:
            import openai  # type: ignore[import]
        except Exception:
            self.provider = None
            return
        api_key = os.getenv("OPENAI_API_KEY")
        if not api_key:
            self.provider = None
            return
        openai.api_key = api_key
        self._openai_client = openai

    def _init_gemini(self) -> None:
        try:
            import google.generativeai as genai  # type: ignore[import]
        except Exception:
            self.provider = None
            return
        api_key = os.getenv("GEMINI_API_KEY")
        if not api_key:
            self.provider = None
            return
        genai.configure(api_key=api_key)
        model_name = os.getenv("GEMINI_MODEL", "gemini-1.5-flash")
        self._gemini_model = genai.GenerativeModel(model_name)

    def legal_chat_answer(self, question: str, context_snippets: List[str]) -> str:
        question = (question or "").strip()
        joined = "\n\n".join(context_snippets[:5])
        prompt = (
            "You are an AI legal information assistant for ordinary citizens. "
            "You are not a lawyer and cannot give official legal advice. "
            "Using ONLY the information in the context and your public knowledge, "
            "answer the user's question in clear, non-technical language, 3–6 short paragraphs.\n\n"
            f"User question:\n{question}\n\n"
            f"Context snippets:\n{joined}\n\n"
            "If you are unsure, say that they should consult a qualified lawyer or legal aid clinic."
        )
        out = self._generate_text(prompt)
        if out:
            return out.strip()
        return (
            "This is general legal information based on similar situations. "
            "For decisions about your specific case, please speak with a qualified lawyer or free legal aid clinic."
        )

    def simplify_clause(self, clause: str) -> str:
        clause = (clause or "").strip()
        if not clause:
            return "I need the clause text to explain it simply."
        prompt = (
            "Explain the following legal clause in very simple language for a 10 year old. "
            "Avoid legal jargon. Keep it under 3 sentences.\n\n"
            f"Clause:\n{clause}"
        )
        out = self._generate_text(prompt)
        if out:
            return out.strip()
        return (
            "This clause talks about who has which responsibilities. "
            "If something goes wrong, it explains who might have to fix the problem or pay for it."
        )

    def multilingual_explanation(self, text: str, language: str) -> str:
        text = (text or "").strip()
        if not text:
            return "No explanation to translate."
        prompt = (
            "Translate the following explanation for ordinary citizens into the "
            f"{language} language. Keep the tone clear, respectful, and simple.\n\n"
            f"Text:\n{text}"
        )
        out = self._generate_text(prompt)
        if out:
            return out.strip()
        return f"[Translation to {language} not available in offline mode]\n\n{text}"

    def _generate_text(self, prompt: str) -> str:
        if not self.provider:
            return ""
        if self.provider == "openai" and self._openai_client:
            try:
                res = self._openai_client.ChatCompletion.create(
                    model=os.getenv("OPENAI_MODEL", "gpt-4o-mini"),
                    messages=[
                        {"role": "system", "content": "You are a helpful legal information assistant."},
                        {"role": "user", "content": prompt},
                    ],
                    temperature=0.3,
                )
                return res.choices[0].message["content"]
            except Exception:
                return ""
        if self.provider == "gemini" and self._gemini_model:
            try:
                res = self._gemini_model.generate_content(prompt)
                return getattr(res, "text", "") or ""
            except Exception:
                return ""
        return ""

llm_client = LLMClient()
""")
```

```python
write(f"{ROOT}/backend/main.py", r"""
from __future__ import annotations
from typing import List, Optional
from fastapi import FastAPI, File, Form, UploadFile
from fastapi.middleware.cors import CORSMiddleware
from pydantic import BaseModel, Field
from complaint_generator import ComplaintRequest, ComplaintType, generate_complaint
from document_analysis import ContractAnalysisResult, analyze_contract_text, extract_text_from_upload
from llm_client import llm_client
from reporting import build_report_payload
from scam_analyzer import ScamAnalysis, analyze_scam_message
from speech_module import transcribe_audio
from utils import safe_str
from vectify_client import retrieve_legal_context
app = FastAPI(title="KnowRights-AI", version="0.4.0")
app.add_middleware(
    CORSMiddleware,
    allow_origins=["*"],
    allow_credentials=True,
    allow_methods=["*"],
    allow_headers=["*"],
)
class ChatRequest(BaseModel):
    message: str = Field(..., min_length=1)
    top_k: int = Field(5, ge=1, le=10)
class ChatResponse(BaseModel):
    answer: str
    context_snippets: List[str]
class ContractAnalysisResponse(BaseModel):
    safety_score: int
    safety_level: str
    clauses: List[dict]
class ScamAnalysisRequest(BaseModel):
    message: str = Field(..., min_length=1)
class ScamAnalysisResponse(BaseModel):
    scam_probability: float
    phishing_type: str
    legal_section: str
    explanation: str
class ClauseExplainRequest(BaseModel):
    clause: str = Field(..., min_length=5)
class MultilingualExplainRequest(BaseModel):
    text: str = Field(..., min_length=5)
    language: str = Field(..., description="Target language name or ISO code, e.g. 'Hindi' or 'hi'")
@app.get("/health")
def health() -> dict:
    return {"ok": True}
@app.post("/chat", response_model=ChatResponse)
def chat_route(req: ChatRequest) -> ChatResponse:
    ctx = retrieve_legal_context(req.message, top_k=req.top_k)
    answer = llm_client.legal_chat_answer(req.message, ctx)
    return ChatResponse(answer=answer, context_snippets=ctx)
@app.post("/analyze-contract", response_model=ContractAnalysisResponse)
async def analyze_contract_endpoint(
    file: Optional[UploadFile] = File(None),
    text: Optional[str] = Form(None),
) -> ContractAnalysisResponse:
    if file is not None:
        content = await extract_text_from_upload(file)
    else:
        content = safe_str(text)
    result: ContractAnalysisResult = analyze_contract_text(content)
    clauses_payload: List[dict] = []
    for c in result.clauses:
        clauses_payload.append(
            {
                "clause": c.clause,
                "risk_score": c.risk_score,
                "risk_class": c.risk_class,
                "explanation": c.explanation,
            }
        )
    return ContractAnalysisResponse(
        safety_score=result.safety_score,
        safety_level=result.safety_level,
        clauses=clauses_payload,
    )
@app.post("/contract-report")
async def contract_report_endpoint(
    file: Optional[UploadFile] = File(None),
    text: Optional[str] = Form(None),
    user_issue: Optional[str] = Form(None),
    suggested_actions: Optional[str] = Form(None),
) -> dict:
    if file is not None:
        content = await extract_text_from_upload(file)
    else:
        content = safe_str(text)
    result: ContractAnalysisResult = analyze_contract_text(content)
    actions_list: List[str] = []
    if suggested_actions:
        actions_list = [a.strip() for a in suggested_actions.split(",") if a.strip()]
    return build_report_payload(
        contract=result,
        user_issue=user_issue or "",
        suggested_actions=actions_list,
    )
@app.post("/analyze-scam", response_model=ScamAnalysisResponse)
def analyze_scam_endpoint(req: ScamAnalysisRequest) -> ScamAnalysisResponse:
    analysis: ScamAnalysis = analyze_scam_message(req.message)
    return ScamAnalysisResponse(
        scam_probability=analysis.scam_probability,
        phishing_type=analysis.phishing_type,
        legal_section=analysis.legal_section,
        explanation=analysis.explanation,
    )
@app.post("/generate-complaint")
def generate_complaint_endpoint(
    complaint_type: ComplaintType = Form(..., description="consumer | police | legal_notice"),
    user_name: str = Form(...),
    opponent_name: str = Form(...),
    facts: str = Form(...),
    location: str = Form(...),
) -> dict:
    req = ComplaintRequest(
        complaint_type=complaint_type,
        user_name=user_name,
        opponent_name=opponent_name,
        facts=facts,
        location=location,
    )
    text = generate_complaint(req)
    return {"complaint": text}
@app.post("/speech-to-text")
async def speech_to_text_endpoint(file: UploadFile = File(...)) -> dict:
    data = await file.read()
    text = transcribe_audio(data)
    return {"text": text}
@app.post("/explain-clause-simple")
def explain_clause_simple(req: ClauseExplainRequest) -> dict:
    explanation = llm_client.simplify_clause(req.clause)
    return {"simple_explanation": explanation}
@app.post("/explain-multilingual")
def explain_multilingual(req: MultilingualExplainRequest) -> dict:
    translated = llm_client.multilingual_explanation(req.text, req.language)
    return {"translated_explanation": translated}
""")
# ensure dirs exist
for d in ["datasets", "models", "notebooks", "logs"]:
    pathlib.Path(f"{ROOT}/{d}").mkdir(parents=True, exist_ok=True)
print("Backend scaffolded at", ROOT)
```
